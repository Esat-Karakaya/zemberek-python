```python
import time
import logging

from zemberek import (
    TurkishSpellChecker,
    TurkishSentenceNormalizer,
    TurkishSentenceExtractor,
    TurkishMorphology,
    TurkishTokenizer
)
```

```python
from utils import get_morphology
morphology = get_morphology()
```

```python
# SINGLE WORD MORPHOLOGICAL ANALYSIS
results = morphology.analyze("ile")
for result in results:
    print(result)
print("\n")
```

```python
# SENTENCE ANALYSIS AND DISAMBIGUATION
sentence = "saat 12.00'da burada olur. Örnek1'e bakın siz."
analysis = morphology.analyze_sentence(sentence)
after = morphology.disambiguate(sentence, analysis)

print("\nAfter disambiguation")
for s in after.best_analysis():
    print(s.format_string())
```
